# IHMValidation Quick Start Guide

This notebook demonstrates basic usage of IHMValidation container.

## Prerequisites
- Container built: `IHMValidation/ihmvalidation_complete.sif`
- Test data available: `test-data-extended/`

```python
import subprocess
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

# Check container exists
container = Path('IHMValidation/ihmvalidation_complete.sif')
assert container.exists(), "Container not found. Run: bash install.sh"
print(f"✓ Container found: {container}")
```

## Run Validation on Example Structure

```python
# Run validation
structure = 'test-data-extended/PDBDEV_00000001.cif'

cmd = [
    'singularity', 'exec',
    'IHMValidation/ihmvalidation_complete.sif',
    'python3', '/opt/IHMValidation/ihm_validation/ihm_validator.py',
    '-f', structure,
    '--output-root', 'notebook_output',
    '--output-prefix', 'demo'
]

print("Running validation...")
result = subprocess.run(cmd, capture_output=True, text=True)

if result.returncode == 0:
    print("✓ Validation completed successfully")
else:
    print(f"Validation output:\n{result.stdout}")
```

## Examine Outputs

```python
# List generated files
output_dir = Path('notebook_output/demo')

if output_dir.exists():
    files = list(output_dir.rglob('*'))
    print(f"Generated {len(files)} files:")
    
    pdf_files = [f for f in files if f.suffix == '.pdf']
    print(f"\nPDF Reports ({len(pdf_files)}):")
    for f in pdf_files:
        print(f"  - {f.name} ({f.stat().st_size / 1024:.1f} KB)")
else:
    print("No output directory found")
```
